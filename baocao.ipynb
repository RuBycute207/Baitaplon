```python
# ==========================================================
# PROJECT: HỆ THỐNG QUẢN LÝ DANH BẠ ĐIỆN TỬ
# MÔN: LẬP TRÌNH PYTHON
# PHẦN 1: KHAI BÁO DỮ LIỆU VÀ CÁC HÀM CƠ BẢN
# ==========================================================


# ==========================================================
# 1. DANH SÁCH LƯU CÁC LIÊN HỆ
# ==========================================================

# Mỗi liên hệ được lưu dưới dạng Dictionary
# Sau đó tất cả liên hệ được lưu trong List

danh_ba = []


# ==========================================================
# 2. KIỂM TRA TÊN KHÔNG ĐƯỢC ĐỂ TRỐNG
# ==========================================================

def nhap_ten():
    while True:
        ten = input("Nhập tên: ").strip()

        if ten == "":
            print("Tên không được để trống!")

        else:
            return ten


# ==========================================================
# 3. NHẬP VÀ KIỂM TRA SỐ ĐIỆN THOẠI
# ==========================================================

def nhap_so_dien_thoai():
    while True:
        sdt = input("Nhập số điện thoại: ").strip()

        if sdt == "":
            print("Số điện thoại không được để trống!")

        elif not sdt.isdigit():
            print("Số điện thoại chỉ được nhập số!")

        elif len(sdt) < 10 or len(sdt) > 11:
            print("Số điện thoại phải có 10 hoặc 11 số!")

        else:
            return sdt


# ==========================================================
# 4. KIỂM TRA EMAIL
# ==========================================================

def nhap_email():
    while True:
        email = input("Nhập Email: ").strip()

        if email == "":
            print("Email không được để trống!")

        elif "@" not in email:
            print("Email phải có ký tự @!")

        elif "." not in email:
            print("Email không đúng định dạng!")

        else:
            return email


# ==========================================================
# 5. KIỂM TRA SỐ ĐIỆN THOẠI ĐÃ TỒN TẠI
# ==========================================================

def kiem_tra_trung_sdt(sdt):
    for lien_he in danh_ba:
        if lien_he["sdt"] == sdt:
            return True

    return False


# ==========================================================
# 6. KIỂM TRA EMAIL ĐÃ TỒN TẠI
# ==========================================================

def kiem_tra_trung_email(email):
    for lien_he in danh_ba:
        if lien_he["email"].lower() == email.lower():
            return True

    return False


# ==========================================================
# 7. TÌM LIÊN HỆ THEO SỐ ĐIỆN THOẠI
# ==========================================================

def tim_lien_he_theo_sdt(sdt):
    for lien_he in danh_ba:
        if lien_he["sdt"] == sdt:
            return lien_he

    return None


# ==========================================================
# 8. HIỂN THỊ MỘT LIÊN HỆ
# ==========================================================

def hien_thi_lien_he(lien_he, stt):
    print(
        f"{stt:<5}"
        f"{lien_he['ten']:<25}"
        f"{lien_he['sdt']:<15}"
        f"{lien_he['email']:<30}"
    )


# ==========================================================
# 9. HIỂN THỊ TIÊU ĐỀ DANH BẠ
# ==========================================================

def hien_thi_tieu_de():
    print("\n")
    print("=" * 80)
    print("                    DANH BẠ ĐIỆN TỬ")
    print("=" * 80)
    print(
        f"{'STT':<5}"
        f"{'TÊN':<25}"
        f"{'SỐ ĐIỆN THOẠI':<15}"
        f"{'EMAIL':<30}"
    )
    print("-" * 80)
    # ==========================================================
# PHẦN 2: THÊM LIÊN HỆ VÀ XEM TOÀN BỘ DANH BẠ
# ==========================================================


# ==========================================================
# 1. THÊM LIÊN HỆ MỚI
# ==========================================================

def them_lien_he():
    print("\n")
    print("=" * 80)
    print("                    THÊM LIÊN HỆ MỚI")
    print("=" * 80)

    # Nhập tên
    ten = nhap_ten()

    # Nhập số điện thoại
    while True:
        sdt = nhap_so_dien_thoai()

        if kiem_tra_trung_sdt(sdt):
            print("Số điện thoại đã tồn tại!")
            print("Vui lòng nhập số điện thoại khác.")
        else:
            break

    # Nhập Email
    while True:
        email = nhap_email()

        if kiem_tra_trung_email(email):
            print("Email đã tồn tại!")
            print("Vui lòng nhập Email khác.")
        else:
            break

    # Tạo một liên hệ mới bằng Dictionary
    lien_he_moi = {
        "ten": ten,
        "sdt": sdt,
        "email": email
    }

    # Thêm liên hệ vào danh sách
    danh_ba.append(lien_he_moi)

    print("\n--------------------------------------------")
    print("Đã thêm liên hệ thành công!")
    print("Tên :", ten)
    print("SĐT :", sdt)
    print("Email:", email)
    print("--------------------------------------------")


# ==========================================================
# 2. XEM TOÀN BỘ DANH BẠ
# ==========================================================

def xem_danh_ba():
    print("\n")

    if len(danh_ba) == 0:
        print("=" * 80)
        print("                    DANH BẠ ĐIỆN TỬ")
        print("=" * 80)
        print("Danh bạ hiện đang trống!")
        print("=" * 80)
        return

    hien_thi_tieu_de()

    # Duyệt qua từng liên hệ trong danh sách
    stt = 1

    for lien_he in danh_ba:
        hien_thi_lien_he(lien_he, stt)
        stt += 1

    print("=" * 80)
    print("Tổng số liên hệ:", len(danh_ba))
    # ==========================================================
# PHẦN 3: TÌM KIẾM VÀ XÓA LIÊN HỆ
# ==========================================================


# ==========================================================
# 1. TÌM KIẾM LIÊN HỆ
# ==========================================================

def tim_kiem_lien_he():
    print("\n")
    print("=" * 80)
    print("                    TÌM KIẾM LIÊN HỆ")
    print("=" * 80)

    if len(danh_ba) == 0:
        print("Danh bạ hiện đang trống!")
        return

    # Nhập từ khóa cần tìm
    tu_khoa = input(
        "Nhập tên, số điện thoại hoặc Email cần tìm: "
    ).strip().lower()

    if tu_khoa == "":
        print("Từ khóa tìm kiếm không được để trống!")
        return

    tim_thay = False
    stt = 1

    print("\nKẾT QUẢ TÌM KIẾM:")
    hien_thi_tieu_de()

    # Duyệt toàn bộ danh bạ
    for lien_he in danh_ba:

        ten = lien_he["ten"].lower()
        sdt = lien_he["sdt"].lower()
        email = lien_he["email"].lower()

        # Tìm gần đúng trong tên hoặc SĐT hoặc Email
        if (
            tu_khoa in ten
            or tu_khoa in sdt
            or tu_khoa in email
        ):
            hien_thi_lien_he(lien_he, stt)
            stt += 1
            tim_thay = True

    print("=" * 80)

    if not tim_thay:
        print("Không tìm thấy liên hệ phù hợp với:", tu_khoa)

    else:
        print("Tìm thấy", stt - 1, "liên hệ.")


# ==========================================================
# 2. XÓA LIÊN HỆ
# ==========================================================

def xoa_lien_he():
    print("\n")
    print("=" * 80)
    print("                     XÓA LIÊN HỆ")
    print("=" * 80)

    if len(danh_ba) == 0:
        print("Danh bạ hiện đang trống!")
        return

    # Nhập số điện thoại của liên hệ cần xóa
    sdt_can_xoa = input(
        "Nhập số điện thoại của liên hệ cần xóa: "
    ).strip()

    if sdt_can_xoa == "":
        print("Số điện thoại không được để trống!")
        return

    # Tìm vị trí của liên hệ
    vi_tri = -1

    for i in range(len(danh_ba)):
        if danh_ba[i]["sdt"] == sdt_can_xoa:
            vi_tri = i
            break

    # Không tìm thấy
    if vi_tri == -1:
        print(
            "Không tìm thấy liên hệ có số điện thoại:",
            sdt_can_xoa
        )
        return

    # Hiển thị liên hệ chuẩn bị xóa
    print("\nLiên hệ tìm thấy:")
    hien_thi_tieu_de()
    hien_thi_lien_he(danh_ba[vi_tri], 1)
    print("=" * 80)

    # Xác nhận xóa
    xac_nhan = input(
        "Bạn có chắc muốn xóa liên hệ này? (y/n): "
    ).strip().lower()

    if xac_nhan == "y":

        # Xóa liên hệ bằng pop()
        lien_he_da_xoa = danh_ba.pop(vi_tri)

        print("\nĐã xóa liên hệ thành công!")
        print("Tên:", lien_he_da_xoa["ten"])

    elif xac_nhan == "n":
        print("Đã hủy thao tác xóa.")

    else:
        print("Lựa chọn không hợp lệ. Đã hủy thao tác xóa.")


# ==========================================================
# 3. HIỂN THỊ THÔNG TIN SAU KHI XÓA
# ==========================================================

def xem_sau_khi_xoa():
    print("\nDanh bạ sau khi xóa:")
    xem_danh_ba()
    # ==========================================================
# PHẦN 4: CHỨC NĂNG MỞ RỘNG VÀ MENU CHÍNH
# ==========================================================


# ==========================================================
# 1. CHỨC NĂNG MỞ RỘNG:
#    THỐNG KÊ LIÊN HỆ THEO ĐUÔI EMAIL
# ==========================================================

def thong_ke_email():
    print("\n")
    print("=" * 80)
    print("                 THỐNG KÊ LIÊN HỆ")
    print("=" * 80)

    if len(danh_ba) == 0:
        print("Danh bạ hiện đang trống!")
        return

    # Dictionary dùng để đếm số lượng email
    thong_ke = {}

    for lien_he in danh_ba:

        email = lien_he["email"].lower()

        # Lấy phần sau dấu @
        if "@" in email:
            ten_mien = email.split("@")[1]

            if ten_mien in thong_ke:
                thong_ke[ten_mien] += 1
            else:
                thong_ke[ten_mien] = 1

    print("Số lượng liên hệ theo tên miền Email:")
    print("-" * 40)

    for ten_mien, so_luong in thong_ke.items():
        print(f"{ten_mien:<25}: {so_luong} liên hệ")

    print("-" * 40)
    print("Tổng số liên hệ:", len(danh_ba))


# ==========================================================
# 2. MENU CHÍNH
# ==========================================================

def display_menu():
    print("\n")
    print("=" * 80)
    print("              HỆ THỐNG QUẢN LÝ DANH BẠ ĐIỆN TỬ")
    print("=" * 80)
    print("1. Thêm liên hệ mới")
    print("2. Xem toàn bộ danh bạ")
    print("3. Tìm kiếm liên hệ")
    print("4. Xóa liên hệ")
    print("5. Thống kê Email")
    print("0. Thoát chương trình")
    print("=" * 80)


# ==========================================================
# 3. HÀM MAIN
# ==========================================================

def main():

    while True:

        display_menu()

        try:
            lua_chon = int(input("Nhập lựa chọn: "))

            if lua_chon == 1:
                them_lien_he()

            elif lua_chon == 2:
                xem_danh_ba()

            elif lua_chon == 3:
                tim_kiem_lien_he()

            elif lua_chon == 4:
                xoa_lien_he()

            elif lua_chon == 5:
                thong_ke_email()

            elif lua_chon == 0:
                print("\n")
                print("=" * 80)
                print("        CẢM ƠN BẠN ĐÃ SỬ DỤNG CHƯƠNG TRÌNH!")
                print("=" * 80)
                break

            else:
                print("Vui lòng chọn từ 0 đến 5!")

        except ValueError:
            print("Lựa chọn phải là số!")

        except Exception as e:
            print("Có lỗi xảy ra:", e)


# ==========================================================
# 4. CHẠY CHƯƠNG TRÌNH
# ==========================================================

if __name__ == "__main__":
    main()
```
